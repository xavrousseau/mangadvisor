# 📈 2. Scraping des Données <a name="2-scraping-des-donnees"></a>

Ce chapitre se concentre sur la récupération des données de l'API AniList. Nous utilisons une requête GraphQL pour extraire des informations détaillées sur les mangas, en exploitant la pagination pour naviguer parmi les pages de résultats.

## 2.1 🛠️ Requête GraphQL <a name="21-requete-graphql"></a>
Préparation de la requête GraphQL pour récupérer les informations sur les mangas.


```python
# Requête GraphQL pour récupérer les informations de chaque manga
query = '''
query ($page: Int, $perPage: Int) {
  Page(page: $page, perPage: $perPage) {
    media(type: MANGA) {
      id
      title {
        romaji
        english
        native
        userPreferred
      }
      format
      status
      description
      startDate {
        year
        month
        day
      }
      endDate {
        year
        month
        day
      }
      chapters
      volumes
      genres
      synonyms
      source
      isAdult
      averageScore
      meanScore
      popularity
      favourites
      countryOfOrigin
      hashtag
      trailer {
        id
        site
        thumbnail
      }
      coverImage {
        extraLarge
        large
        medium
        color
      }
      bannerImage
      staff {
        edges {
          node {
            id
            name {
              full
              native
            }
          }
        }
      }
      characters {
        edges {
          node {
            id
            name {
              full
              native
            }
          }
        }
      }
      externalLinks {
        site
        url
      }
      tags {
        id
        name
        description
        category
        rank
        isGeneralSpoiler
        isMediaSpoiler
        isAdult
      }
    }
    pageInfo {
      hasNextPage
      currentPage
    }
  }
}
'''
```

## 2.2 Fonction `send_request` <a name="22-fonction-send-request"></a>

Cette fonction envoie une requête HTTP POST à l'API AniList pour récupérer des données de mangas, en utilisant la requête GraphQL définie plus haut.

- **Arguments** :
  - `query` : La requête GraphQL contenant les instructions de récupération des données.
  - `variables` : Un dictionnaire avec les paramètres de pagination (`page` et `perPage`) pour contrôler la page et le nombre de résultats.

- **Retour** : 
  - La fonction retourne les données JSON en cas de succès, ou `None` si toutes les tentatives échouent.

- **Gestion des erreurs** :
  - La fonction effectue plusieurs tentatives (`max_attempts`), avec un délai (`wait_time`) entre chaque tentative si une erreur de requête est levée.
  - En cas de succès, elle logue un message de confirmation ; en cas d'échec, elle logue l'erreur pour chaque tentative et une erreur finale si toutes les tentatives échouent.


```python
def send_request(query, variables):
    """
    Envoie une requête POST avec la requête GraphQL et gère les erreurs avec tentatives multiples.

    Args:
        query (str): La requête GraphQL.
        variables (dict): Variables pour la pagination.

    Returns:
        dict | None: Données JSON en cas de succès, None sinon.
    """
    # Boucle de tentatives de requête pour gérer les erreurs de connexion
    for attempt in range(1, notebook_config['max_attempts'] + 1):
        try:
            # Envoi de la requête POST à l'API AniList
            response = requests.post(notebook_config['api_url'], json={'query': query, 'variables': variables})
            response.raise_for_status()  # Lève une exception si la réponse est invalide (e.g., 404 ou 500)

            # Log de la réussite de la requête et retour des données JSON
            logger.info(f"Requête réussie pour la page {variables['page']}.")
            return response.json()
        
        except requests.exceptions.RequestException as e:
            # En cas d'erreur, log et attente avant la prochaine tentative
            logger.error(f"Tentative {attempt} échouée pour la page {variables['page']}: {e}")
            if attempt < notebook_config['max_attempts']:  # Attendre avant une nouvelle tentative
                time.sleep(notebook_config["wait_time"])
            else:
                logger.error("Échec définitif de la requête après plusieurs tentatives.")
    
    # Retourne None en cas d'échec après toutes les tentatives
    return None
```

## 2.3 🚀 Exécution du Scraping avec Pagination <a name="23-execution-du-scraping-avec-pagination"></a>
Utilisation de la pagination pour récupérer toutes les pages de résultats


# Fonction `get_all_mangas`

Cette fonction effectue le scraping complet des mangas disponibles via l'API AniList en utilisant la pagination.

- **Description** :
  - La fonction commence à la page 1 et utilise une boucle pour récupérer chaque page de données en envoyant une requête via la fonction `send_request`.
  - Elle stocke toutes les données récupérées dans une liste `all_mangas`.

- **Détails de la logique** :
  - Chaque page de résultats est ajoutée à la liste `all_mangas`.
  - La fonction vérifie la présence d'une page suivante avec `hasNextPage`. Si une page suivante est disponible, elle incrémente la page actuelle pour continuer le scraping.
  - Si la fonction détecte qu’il n'y a plus de pages disponibles, elle termine la boucle et renvoie `all_mangas`.

- **Gestion des erreurs** :
  - En cas d’erreur de requête (si `send_request` retourne `None`), la fonction enregistre un message d'erreur et arrête le processus.

- **Retour** :
  - La fonction renvoie `all_mangas`, une liste contenant toutes les données de mangas récupérées de l'API AniList.


```python
def get_all_mangas():
    """
    Récupère toutes les pages de mangas en utilisant la pagination de l'API AniList avec une barre de progression.

    Returns:
        list: Liste de tous les mangas récupérés.
    """
    logger.info("Début du scraping de la base de données de mangas.")
    all_mangas = []  # Liste pour stocker tous les mangas récupérés
    page = 1  # Page de départ

    # Récupération initiale pour obtenir le nombre total de pages
    variables = {'page': page, 'perPage': notebook_config['per_page']}
    initial_data = send_request(query, variables)

    if initial_data:
        # Calcul du nombre total de pages (si possible)
        total_pages = initial_data['data']['Page']['pageInfo']['currentPage'] + \
                      (1 if initial_data['data']['Page']['pageInfo']['hasNextPage'] else 0)
        logger.info(f"Nombre total estimé de pages à scraper : {total_pages}")
    else:
        logger.error("Impossible d'obtenir le nombre total de pages. Arrêt du processus.")
        return []

    # Utilisation de tqdm pour suivre la progression
    with tqdm(total=total_pages, desc="Scraping pages AniList", unit="page") as pbar:
        while True:
            # Variables de pagination pour chaque requête
            variables = {'page': page, 'perPage': notebook_config['per_page']}
            data = send_request(query, variables)

            if data:
                # Ajout des mangas de la page actuelle à la liste
                all_mangas.extend(data['data']['Page']['media'])
                logger.info(f"Page {page} récupérée avec succès.")

                # Vérification de l'existence d'une page suivante
                if data['data']['Page']['pageInfo']['hasNextPage']:
                    page += 1  # Passer à la page suivante
                    time.sleep(notebook_config['wait_time'])  # Pause pour respecter les limites d'API
                else:
                    logger.info("Toutes les pages ont été récupérées.")
                    break  # Fin de la récupération si plus de pages

                # Met à jour la barre de progression
                pbar.update(1)
            else:
                logger.error("Erreur lors de la récupération des données. Arrêt du processus.")
                break  # Arrêt en cas d'erreur

    return all_mangas
```

## 2.4 📁 Sauvegarde des Données dans Excel <a name="24-sauvegarde-des-donnees-dans-excel"></a>
Transformation des données récupérées en un DataFrame et exportation dans un fichier Excel.


# Fonction `save_mangas_to_excel`

Cette fonction transforme la liste de mangas récupérée en un DataFrame, puis l’exporte dans un fichier Excel.

- **Arguments** :
  - `mangas` : Une liste de dictionnaires contenant les informations de chaque manga, telle que renvoyée par la fonction `get_all_mangas`.

- **Fonctionnement** :
  - La fonction commence par transformer chaque entrée de `mangas` en un dictionnaire structuré. Elle extrait des informations spécifiques comme l’ID, les titres, le format, le statut, la date de début et de fin, les genres, la popularité, etc.
  - Les genres et synonymes sont convertis en chaînes de texte séparées par des virgules pour faciliter leur lecture et leur manipulation.

- **Création du DataFrame** :
  - La liste de dictionnaires `manga_list` est convertie en un DataFrame avec `pd.DataFrame()`, ce qui permet de manipuler les données plus facilement.

- **Export en Excel** :
  - Le DataFrame est exporté vers un fichier Excel, dont le chemin est défini dans `config["data_directory"]`.
  - La fonction logue un message de confirmation une fois l'export terminé.

- **Retour** :
  - La fonction ne retourne rien ; elle se contente de sauvegarder les données dans un fichier Excel pour des analyses ultérieures.


```python
def save_mangas_to_csv(mangas):
    """
    Transforme la liste de mangas en DataFrame et l'exporte en Excel.

    Args:
        mangas (list): Liste de dictionnaires contenant les informations de chaque manga.
    """
    # Transformation en liste de dictionnaires structurés
    manga_list = [
        {
            'ID': manga['id'],
            'Romaji Title': manga['title'].get('romaji'),
            'English Title': manga['title'].get('english'),
            'Native Title': manga['title'].get('native'),
            'User Preferred Title': manga['title'].get('userPreferred'),
            'Format': manga.get('format'),
            'Status': manga.get('status'),
            'Description': manga.get('description'),
            'Start Date': manga.get('startDate'),
            'End Date': manga.get('endDate'),
            'Chapters': manga.get('chapters'),
            'Volumes': manga.get('volumes'),
            'Genres': ', '.join(manga['genres']) if manga['genres'] else None,
            'Synonyms': ', '.join(manga['synonyms']) if manga['synonyms'] else None,
            'Source': manga.get('source'),
            'Is Adult': manga.get('isAdult'),
            'Average Score': manga.get('averageScore'),
            'Mean Score': manga.get('meanScore'),
            'Popularity': manga.get('popularity'),
            'Favourites': manga.get('favourites'),
            'Country of Origin': manga.get('countryOfOrigin'),
            'Hashtag': manga.get('hashtag'),
            'Trailer': manga.get('trailer'),
            'Cover Image': manga.get('coverImage'),
            'Banner Image': manga.get('bannerImage'),
            'Staff': manga.get('staff'),
            'Characters': manga.get('characters'),
            'External Links': manga.get('externalLinks'),
            'Tags': manga.get('tags')
        }
        for manga in mangas
    ]

    # Créer un DataFrame à partir de la liste de dictionnaires
    df = pd.DataFrame(manga_list)

    # Définir le chemin de sauvegarde pour le fichier CSV
    output_path = notebook_config["data_directory"] / "mangas_anilist.csv"

    # Exporter le DataFrame dans un fichier CSV
    df.to_csv(output_path, index=False)
    logger.info(f"Données sauvegardées dans le fichier : {output_path}")

```

Dans cette étape finale, nous lançons le processus de scraping pour récupérer toutes les données de mangas et les enregistrer dans un fichier Excel.

# 📊 3. Nettoyage des Données <a name="3-nettoyage-des-donnees"></a>

Dans cette section, nous allons préparer les données pour une analyse exploratoire en nettoyant et en transformant les informations brutes obtenues lors du scraping. ## 3.1 📅 Extraction des Composants de Date <a name="31-extraction-des-composants-de-date"></a>

**Objectif** : Extraire les composants `year`, `month`, et `day` des colonnes `Start Date` et `End Date`, afin de faciliter l'analyse des périodes de publication des mangas.

Nous allons transformer les colonnes de date encapsulées dans des objets JSON en plusieurs colonnes représentant respectivement l'année, le mois et le jour.


```python
# Fonction pour extraire les composants de date
def extraire_composant_date(colonne_date, composant):
    return colonne_date.apply(lambda x: eval(x).get(composant) if isinstance(x, str) else np.nan)

# Extraction des composants 'year', 'month', 'day' pour les colonnes de date
df_mangas['Start Year'] = extraire_composant_date(df_mangas['Start Date'], 'year')
df_mangas['Start Month'] = extraire_composant_date(df_mangas['Start Date'], 'month')
df_mangas['Start Day'] = extraire_composant_date(df_mangas['Start Date'], 'day')
df_mangas['End Year'] = extraire_composant_date(df_mangas['End Date'], 'year')
df_mangas['End Month'] = extraire_composant_date(df_mangas['End Date'], 'month')
df_mangas['End Day'] = extraire_composant_date(df_mangas['End Date'], 'day')
```

## 3.2 🧹 Nettoyage des Données <a name="32-nettoyage-des-donnees"></a>

**Objectif** : Supprimer les colonnes inutiles, gérer les valeurs manquantes, et préparer les données pour l'analyse.

Nous allons supprimer les colonnes non pertinentes, gérer les valeurs manquantes, et convertir les dates en formats exploitables pour l'analyse.

```python
# Suppression des colonnes non pertinentes
df_mangas.drop(columns=['Start Date', 'End Date', 'Cover Image', 'Tags', 'Synonyms', 'Hashtag', 'Trailer', 'Characters', 'Staff', 'Banner Image', 'External Links'], inplace=True)

# Remplacer les valeurs NaN par None pour permettre la création des dates
df_mangas[['Start Year', 'Start Month', 'Start Day']] = df_mangas[['Start Year', 'Start Month', 'Start Day']].replace({np.nan: None})
df_mangas[['End Year', 'End Month', 'End Day']] = df_mangas[['End Year', 'End Month', 'End Day']].replace({np.nan: None})

# Créer une colonne 'Start Date' en combinant 'Start Year', 'Start Month', et 'Start Day'
df_mangas['Start Date'] = df_mangas.apply(
    lambda row: pd.to_datetime(f"{int(row['Start Year'])}-{int(row['Start Month']):02d}-{int(row['Start Day']):02d}", errors='coerce')
    if row['Start Year'] and row['Start Month'] and row['Start Day'] else pd.NaT, axis=1
)

# Créer une colonne 'End Date' en combinant 'End Year', 'End Month', et 'End Day'
df_mangas['End Date'] = df_mangas.apply(
    lambda row: pd.to_datetime(f"{int(row['End Year'])}-{int(row['End Month']):02d}-{int(row['End Day']):02d}", errors='coerce')
    if row['End Year'] and row['End Month'] and row['End Day'] else pd.NaT, axis=1
)

# Conversion des dates en chaînes de caractères formatées, laissant les NaT vides
df_mangas['Start Date'] = df_mangas['Start Date'].dt.strftime('%Y-%m-%d').fillna('')
df_mangas['End Date'] = df_mangas['End Date'].dt.strftime('%Y-%m-%d').fillna('')

# Suppression des colonnes non pertinentes
df_mangas.drop(columns=['Start Year', 'Start Month', 'Start Day','End Year', 'End Month', 'End Day'], inplace=True)
```


## 📊 4. Analyse Exploratoire des Données <a name="Analyse-Exploratoire-des-Données"></a>


```python
def description_donnees(dataframe: pd.DataFrame) -> pd.DataFrame:
    """
    Analyse un DataFrame et retourne un résumé sous forme de DataFrame incluant diverses statistiques.
    
    Arguments:
    dataframe (pd.DataFrame) : Le DataFrame à analyser.
    
    Retourne:
    pd.DataFrame : Un DataFrame contenant le résumé des informations analysées.
    """
    
    # Convertit toutes les colonnes en chaînes de caractères pour éviter les erreurs 
    # potentielles liées aux types non hashables (comme les listes ou dictionnaires)
    dataframe_str = dataframe.astype(str)

    # Calcul du nombre total de lignes et de colonnes dans le DataFrame
    nb_lignes, nb_colonnes = dataframe.shape

    # Calcul du pourcentage total de valeurs manquantes (NaN) dans le DataFrame
    pourcentage_nan = round(dataframe.isna().sum().sum() / dataframe.size * 100, 2)

    # Calcul du nombre de lignes en double
    nombre_doublons = dataframe_str.duplicated().sum()
    pourcentage_doublons = round(nombre_doublons / nb_lignes * 100, 2) if nb_lignes > 0 else 0

    # Comptage des types de colonnes dans le DataFrame
    nb_colonnes_objet = dataframe.select_dtypes(include=['object']).shape[1]
    nb_colonnes_flottant = dataframe.select_dtypes(include=['float']).shape[1]
    nb_colonnes_entier = dataframe.select_dtypes(include=['int']).shape[1]
    nb_colonnes_booleen = dataframe.select_dtypes(include=['bool']).shape[1]

    # Taille mémoire utilisée par le DataFrame en Mo
    taille_memoire = round(dataframe.memory_usage().sum() / 1024**2, 3)

    # Vue d'ensemble avec skimpy (pour générer des statistiques détaillées)
    print("Résumé du DataFrame avec skimpy:")
    skim_result = skim(dataframe)  # Affiche un résumé général du DataFrame
    print(skim_result)

    # Résumé des informations calculées dans un dictionnaire
    info_dataframe = {
        'Lignes': nb_lignes,
        'Colonnes': nb_colonnes,
        '% NaN': pourcentage_nan,
        '% Doublons': pourcentage_doublons,
        'Colonnes Objet': nb_colonnes_objet,
        'Colonnes Flottant': nb_colonnes_flottant,
        'Colonnes Entier': nb_colonnes_entier,
        'Colonnes Booléen': nb_colonnes_booleen,
        'Mémoire (Mo)': taille_memoire
    }

    # Convertit le dictionnaire en DataFrame pour une présentation claire sous forme de tableau
    return pd.DataFrame(info_dataframe, index=[0])
```

```python
# Appel de la fonction d'analyse descriptive
resultat_description = description_donnees(df_mangas)

# Affichage du résultat de l'analyse descriptive
print("Résumé descriptif du DataFrame :")
print(resultat_description)
```


### 4.2 Fréquence des Publications par Année <a name="Frequence-des-Publications-par-Annee"></a>

Exploration de la fréquence des publications par année pour détecter les tendances de production dans l'industrie du manga.


```python

# Extraire l'année des dates de publication
df_mangas['Start Year'] = pd.to_datetime(df_mangas['Start Date'], errors='coerce').dt.year

# Comptage des mangas publiés par année de début
start_year_counts = df_mangas['Start Year'].value_counts().sort_index()
fig2 = px.line(start_year_counts, x=start_year_counts.index, y=start_year_counts.values, 
               title="Fréquence de Publication des Mangas par Année de Début")
fig2.update_layout(xaxis_title="Année", yaxis_title="Nombre de Publications", template="plotly_dark")
fig2.show()
```


### 4.3 Répartition des Genres <a name="Repartition-des-Genres"></a>

Identification des genres dominants dans notre dataset de mangas pour comprendre les préférences de genre.


```python

# Remplir les genres manquants pour une meilleure lisibilité
df_mangas['Genres'] = df_mangas['Genres'].fillna("Unknown")
genres_expanded = df_mangas['Genres'].str.split(', ').explode()
genre_counts = genres_expanded.value_counts()

# Tracer la distribution des genres
fig3 = px.bar(genre_counts, x=genre_counts.index, y=genre_counts.values, 
              title="Répartition des Genres des Mangas")
fig3.update_layout(xaxis_title="Genre", yaxis_title="Fréquence", template="plotly_dark")
fig3.show()
```


### 4.4 Analyse de la Popularité et des Favoris <a name="Analyse-de-la-Popularite-et-des-Favoris"></a>

Exploration des mangas les plus populaires et ceux ayant le plus de favoris pour comprendre les préférences des utilisateurs.


```python

fig = px.scatter(df_mangas, x="Popularity", y="Favourites", title="Popularité vs Favoris",
                 hover_data=['Romaji Title'], labels={'Popularity': 'Popularité', 'Favourites': 'Favoris'})
fig.update_layout(template="plotly_dark")
fig.show()
```


### 4.5 Proportion de Contenu Adulte (NSFW) <a name="Proportion-de-Contenu-Adulte"></a>

Examen de la proportion de mangas destinés à un public adulte et comparaison de leur popularité.


```python

# Distribution du contenu NSFW
nsfw_counts = df_mangas['Is Adult'].value_counts()
fig4 = px.pie(nsfw_counts, names=nsfw_counts.index.map({True: 'Contenu Adulte', False: 'Contenu Général'}), 
              values=nsfw_counts.values, title="Proportion de Contenu Adulte vs Contenu Général")
fig4.update_traces(hole=0.4)
fig4.update_layout(template="plotly_dark")
fig4.show()
```
