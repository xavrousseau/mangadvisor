# 3. 🛠️ Définition des fonctions principales

Dans cette section, nous définissons les fonctions essentielles pour accomplir les tâches de scraping, de gestion des erreurs et de sauvegarde des données. Ces fonctions rendent le code plus modulaire, réutilisable et maintenable.

## 📚 Fonctions Utilitaires

Ces fonctions jouent un rôle clé pour garantir la robustesse et la fiabilité du script.

---

### **`fetch_manga_details`**
- **Description :** Récupère les détails d'un manga à partir de son URL en utilisant la bibliothèque `requests` pour envoyer une requête HTTP et `BeautifulSoup` pour extraire les données HTML.
- **Détails récupérés :**
  - Titre
  - Titre VO
  - Titre traduit
  - Résumé
  - Origine
- **Gestion des erreurs :**
  - Si une erreur réseau survient (par exemple, une requête échoue), elle est journalisée avec un message explicite.
  - Chaque champ utilise une valeur par défaut en cas de données manquantes pour éviter les erreurs.

---

### **`fetch_manga_details_with_retry`**
- **Description :** Étend la fonction `fetch_manga_details` en ajoutant une logique de réessai pour améliorer la résilience en cas d'échec temporaire des requêtes.
- **Caractéristiques :**
  - Limite configurable pour le nombre de tentatives (`RETRIES`).
  - Pause entre les tentatives pour éviter de surcharger le serveur cible (`PAUSE_BETWEEN_REQUESTS`).
- **Avantages :**
  - Réduit les risques d'échec complet en cas de problèmes temporaires de réseau.
  - Les échecs répétés sont clairement journalisés pour permettre une analyse ultérieure.

---

### **`save_dataframe_to_csv`**
- **Description :** Sauvegarde un DataFrame Pandas dans un fichier CSV.
- **Caractéristiques :**
  - Vérifie si le DataFrame est vide avant de tenter une sauvegarde, ce qui évite de générer des fichiers inutiles.
  - Enregistre les données avec l'encodage UTF-8 pour garantir la compatibilité.
- **Gestion des erreurs :**
  - Les échecs d'écriture dans le fichier (par exemple, permissions insuffisantes) sont journalisés avec précision.

---

### Pourquoi ces fonctions ?
- **Centralisation :** Chaque fonction est spécialisée dans une tâche unique, ce qui réduit la complexité du code principal.
- **Robustesse :** Les mécanismes de gestion des erreurs et de réessai augmentent la fiabilité du script.
- **Clarté :** Les noms explicites et les journaux détaillés facilitent la compréhension et le débogage.

Avec ces fonctions, nous posons les bases d'un pipeline de scraping robuste, prêt à collecter et organiser des données sur les mangas.


````python
def fetch_manga_details(url: str) -> tuple[str, str, str, str, str]:
    """
    Récupère les détails d'un manga depuis une URL.
    
    Args:
        url (str): URL de la page du manga.
    
    Returns:
        tuple: Contient :
            - Le titre (str)
            - Le titre VO (str)
            - Le titre traduit (str)
            - Le résumé (str)
            - L'origine (str)
    """
    try:
        # Envoyer une requête GET à l'URL spécifiée avec un timeout de 10 secondes
        response = requests.get(url, timeout=10)
        response.raise_for_status()  # Vérifie si la requête a échoué (4xx ou 5xx)

        # Analyse du contenu HTML de la page
        soup = BeautifulSoup(response.content, 'html.parser')

        # Extraction des données avec des sélecteurs CSS
        # Utilisation d'une valeur par défaut en cas d'éléments manquants
        title = soup.select_one('h1.entry-page-title').text.strip() if soup.select_one('h1.entry-page-title') else DEFAULT_ERROR
        original_title = soup.select_one('li.title-vo').text.replace('Titre VO:', '').strip() if soup.select_one('li.title-vo') else DEFAULT_ERROR
        translated_title = soup.select_one('li.trad').text.replace('Titre traduit:', '').strip() if soup.select_one('li.trad') else DEFAULT_ERROR
        summary = ''.join([line.strip() for line in soup.select_one('div.bigsize').stripped_strings]) if soup.select_one('div.bigsize') else DEFAULT_ERROR
        origin = soup.find('strong', string="Origine").find_next_sibling(string=True).strip() if soup.find('strong', string="Origine") else DEFAULT_ERROR

        # Retourne les données extraites sous forme de tuple
        return title, original_title, translated_title, summary, origin

    except requests.RequestException as e:
        # Journaliser une erreur si la requête échoue
        logger.error(f"Erreur lors de la récupération des détails pour {url}: {e}")
        # Retourne des messages d'erreur spécifiques pour chaque champ
        return (CONNECTION_ERROR,) * 5


def fetch_manga_details_with_retry(url: str, retries: int = RETRIES) -> tuple[str, str, str, str, str]:
    """
    Récupère les détails d'un manga avec une logique de réessai en cas d'échec.
    
    Args:
        url (str): URL de la page du manga.
        retries (int): Nombre de tentatives autorisées.
    
    Returns:
        tuple: Contient les détails du manga ou des erreurs si toutes les tentatives échouent.
    """
    for attempt in range(retries):  # Répéter pour le nombre de tentatives spécifié
        details = fetch_manga_details(url)  # Appeler la fonction principale
        if CONNECTION_ERROR not in details:  # Vérifie si la récupération a réussi
            return details  # Retourner les détails si tout va bien

        # Journaliser un avertissement en cas d'échec de la tentative
        logger.warning(f"Tentative {attempt + 1} échouée pour {url}. Réessai dans {PAUSE_BETWEEN_REQUESTS} secondes...")
        time.sleep(PAUSE_BETWEEN_REQUESTS)  # Pause avant la prochaine tentative

    # Si toutes les tentatives échouent, journaliser une erreur et retourner un message d'échec final
    logger.error(f"Échec définitif pour {url} après {retries} tentatives.")
    return (FINAL_ERROR,) * 5


def save_dataframe_to_csv(df: pd.DataFrame, filename: str):
    """
    Sauvegarde un DataFrame dans un fichier CSV.
    
    Args:
        df (pd.DataFrame): DataFrame à sauvegarder.
        filename (str): Chemin complet du fichier CSV.
    
    Notes:
        Vérifie si le DataFrame est vide avant de tenter une sauvegarde.
    """
    # Vérifie si le DataFrame est vide pour éviter de sauvegarder des données inutiles
    if df.empty:
        logger.warning(f"Tentative de sauvegarder un DataFrame vide dans {filename}. Opération annulée.")
        return

    try:
        # Sauvegarder le DataFrame au format CSV avec encodage UTF-8
        df.to_csv(filename, index=False, encoding='utf-8')
        logger.info(f"Fichier sauvegardé : {filename}")
    except IOError as e:
        # Journaliser une erreur si la sauvegarde échoue
        logger.error(f"Erreur lors de la sauvegarde du fichier {filename}: {e}")
````

# 4. 🌐 Scraping des mangas

Cette étape est le cœur du projet, où nous effectuons le scraping systématique des données de mangas à partir de **Manga-News**.

## Objectifs principaux :
1. **Parcourir l'alphabet et le symbole `#` :** Identifier toutes les séries disponibles sur le site, lettre par lettre.
2. **Extraire les informations principales :** Récupérer les données affichées directement sur les pages de listing (titre, auteur, éditeur, etc.).
3. **Obtenir des détails supplémentaires :** Pour chaque série, visiter sa page dédiée et extraire des informations plus précises (titre VO, résumé, origine, etc.) à l'aide de la fonction `fetch_manga_details_with_retry`.
4. **Sauvegarder les résultats :** Organiser les données dans des fichiers CSV distincts pour chaque lettre ou symbole.

## Étapes du processus :
### 1️⃣ Extraction des données des pages de listing :
- Envoi de requêtes HTTP vers les pages de chaque lettre.
- Analyse du HTML pour extraire les informations principales.
- Gestion des erreurs pour éviter les interruptions en cas de requête échouée.

### 2️⃣ Traitement des détails individuels :
- Filtrer les séries avec des URLs valides.
- Collecter les informations complémentaires via des visites répétées, avec une logique de réessai (retry).

### 3️⃣ Sauvegarde des résultats :
- Conversion des données en **DataFrame Pandas**.
- Sauvegarde des données dans des fichiers CSV organisés par lettre.

### 4️⃣ Gestion des pauses :
- Implémentation de pauses entre les requêtes pour éviter de surcharger le serveur, conformément aux bonnes pratiques de scraping.

## Points clés :
- **Robustesse :** Chaque lettre est traitée indépendamment, ce qui permet de continuer même en cas d'erreurs pour une ou plusieurs lettres.
- **Modularité :** Utilisation de fonctions spécifiques pour la récupération des détails et la sauvegarde, rendant le code lisible et maintenable.
- **Conformité :** Respect des limites du serveur avec des pauses configurables via la variable `PAUSE_BETWEEN_REQUESTS`.

Ce processus garantit une collecte méthodique et fiable des données tout en assurant une flexibilité en cas d'interruptions ou de mises à jour du site.


````python
# 🌐 Boucle principale pour parcourir les lettres de l'alphabet et le symbole "#"
# Cette boucle parcourt toutes les lettres de l'alphabet et le caractère spécial '#'
for letter in tqdm(list("ABCDEFGHIJKLMNOPQRSTUVWXYZ") + ['#'], desc="Scraping lettres"):
    # Log pour indiquer la lettre ou le symbole en cours de traitement
    logger.info(f"Traitement de la lettre/symbole : {letter}")
    
    # Définir l'URL correspondant à la page de listing pour la lettre/symbole actuel
    listing_url = f'https://www.manga-news.com/index.php/series/{letter}'
    
    # Liste temporaire pour stocker les informations de base des mangas
    data = []

    # 🕵️‍♂️ Étape 1 : Récupération des mangas de la page de listing
    try:
        # Log pour indiquer l'envoi de la requête HTTP
        logger.info(f"Envoi de la requête vers {listing_url}")
        
        # Requête HTTP pour obtenir la page HTML de la lettre/symbole actuel
        response = requests.get(listing_url, timeout=10)  # Timeout défini pour éviter les requêtes bloquantes
        response.raise_for_status()  # Vérifie si la requête s'est bien déroulée (code 200)

        # Analyse du contenu HTML avec BeautifulSoup pour extraire les informations nécessaires
        soup = BeautifulSoup(response.content, 'html.parser')

        # Sélection des lignes de la table contenant les mangas
        manga_rows = soup.select('tr')

        # 🔄 Boucle sur chaque ligne pour extraire les informations principales
        for row in manga_rows:
            # Extraction des différents éléments HTML contenant les informations sur chaque manga
            title_element = row.select_one('span.item-list-content-title')  # Titre principal
            author_element = row.select_one('td.list-1-data a[title="Dessinateur"]')  # Auteur
            editor_element = row.select('td.list-1-data')[1] if len(row.select('td.list-1-data')) > 1 else None  # Éditeur
            volumes_element = row.select('td.list-1-data')[2] if len(row.select('td.list-1-data')) > 2 else None  # Nombre de volumes
            genre_element = row.select('td.list-1-data')[3] if len(row.select('td.list-1-data')) > 3 else None  # Genre
            link_element = row.select_one('a[title]')  # Lien vers la page du manga

            # Ajouter les informations dans la liste `data` sous forme de dictionnaire
            data.append({
                'Titre': title_element.text.strip() if title_element else "Titre non trouvé",
                'Auteur': author_element.text.strip() if author_element else "Auteur non trouvé",
                'Éditeur': editor_element.text.strip() if editor_element else "Éditeur non trouvé",
                'Nombre de volumes': volumes_element.text.strip() if volumes_element else "Volumes non trouvés",
                'Genre': genre_element.text.strip() if genre_element else "Genre non trouvé",
                'URL': link_element['href'] if link_element else "Lien non trouvé"
            })

        # Log pour indiquer combien de mangas ont été trouvés pour la lettre actuelle
        logger.info(f"{len(data)} mangas trouvés pour la lettre {letter}")

    except requests.RequestException as e:
        # Gestion des erreurs de requête HTTP
        logger.error(f"Erreur lors du traitement de {listing_url}: {e}")
        continue  # Passer à la prochaine lettre en cas d'erreur

    # 🧹 Étape 2 : Conversion des données en DataFrame et traitement des détails
    # Création d'un DataFrame Pandas à partir de la liste de dictionnaires `data`
    df = pd.DataFrame(data)
    if df.empty:
        # Si aucune donnée n'a été récupérée pour la lettre, afficher un avertissement et passer à la suivante
        logger.warning(f"Aucune donnée trouvée pour {letter}.")
        continue

    # Filtrer les lignes contenant des URLs valides (commençant par "https")
    valid_rows = df[df['URL'].str.startswith("https")].copy()

    # 🔍 Récupération des détails supplémentaires pour chaque manga
    try:
        # Log pour indiquer le début de la récupération des détails
        logger.info(f"Récupération des détails pour {len(valid_rows)} mangas")
        
        # Liste pour stocker les résultats détaillés
        details = []

        # Utilisation de tqdm pour afficher une barre de progression pendant la récupération des détails
        for url in tqdm(valid_rows['URL'], desc=f"Récupération détails ({letter})"):
            # Appel de la fonction `fetch_manga_details_with_retry` pour chaque URL
            details.append(fetch_manga_details_with_retry(url))

        # Ajouter les détails récupérés au DataFrame
        valid_rows[['Titre Série', 'Titre VO', 'Titre traduit', 'Résumé', 'Origine']] = pd.DataFrame(details)

    except Exception as e:
        # Gestion des erreurs lors de la récupération des détails
        logger.error(f"Erreur lors de la récupération des détails pour {letter}: {e}")
        continue

    # 💾 Étape 3 : Sauvegarde des résultats dans un fichier CSV
    # Définir le chemin du fichier de sauvegarde pour les résultats de la lettre actuelle
    output_file = CONFIG["processed_directory"] / f'mangas_{letter}.csv'
    try:
        # Sauvegarde du DataFrame dans un fichier CSV
        valid_rows.to_csv(output_file, index=False, encoding="utf-8")
        logger.info(f"Fichier sauvegardé : {output_file}")
    except Exception as e:
        # Gestion des erreurs lors de la sauvegarde
        logger.error(f"Erreur lors de la sauvegarde du fichier {output_file}: {e}")
        continue

    # 💤 Pause entre les lettres pour éviter de surcharger le serveur
    time.sleep(PAUSE_BETWEEN_REQUESTS)

# 📜 Log final
# Indique que le script a terminé son exécution avec succès
logger.info("Tous les fichiers ont été traités avec succès. Le script a terminé son exécution.")
````

# 5. 🔗 Fusion des fichiers CSV

Une fois le scraping terminé pour chaque lettre ou symbole, il est nécessaire de combiner les résultats dans un fichier unique pour faciliter les analyses ultérieures.

## Fonction `merge_all_csv`

### Objectif :
Cette fonction regroupe tous les fichiers CSV générés dans un répertoire en un seul fichier.

### Étapes :
1. **Recherche des fichiers :**
   - Parcourt le répertoire spécifié pour trouver tous les fichiers avec l'extension `.csv`.
   - Vérifie qu'au moins un fichier est disponible.

2. **Lecture et fusion :**
   - Charge chaque fichier en tant que DataFrame Pandas.
   - Fusionne tous les DataFrames en un seul.

3. **Sauvegarde :**
   - Enregistre le DataFrame fusionné dans un nouveau fichier CSV (`mangas_all.csv` par défaut).

### Points clés :
- **Robustesse :** La fonction gère les erreurs de lecture pour chaque fichier individuel, en continuant avec les fichiers valides.
- **Logs détaillés :** Fournit des informations sur les fichiers trouvés, les erreurs rencontrées, et les dimensions du fichier fusionné.
- **Flexibilité :** Le nom du fichier fusionné peut être configuré via le paramètre `final_filename`.

### Appel de la fonction :
```python
merge_all_csv(output_dir)


````python
def merge_all_csv(output_dir, final_filename="mangas_all.csv"):
    """
    Fusionne tous les fichiers CSV individuels présents dans un répertoire en un seul fichier.

    Args:
        output_dir (str): Répertoire contenant les fichiers CSV individuels générés par le scraping.
        final_filename (str): Nom du fichier CSV final fusionné. Par défaut, "mangas_all.csv".

    Notes:
        Vérifie la présence de fichiers CSV dans le répertoire et fusionne uniquement les fichiers valides.
    """
    try:
        # 🗂️ Étape 1 : Récupérer tous les fichiers CSV dans le répertoire spécifié
        all_files = [
            os.path.join(output_dir, f)  # Construit le chemin complet des fichiers
            for f in os.listdir(output_dir)  # Parcourt les fichiers du répertoire
            if f.endswith('.csv')  # Filtre uniquement les fichiers avec extension '.csv'
        ]
        
        # Vérifie si des fichiers CSV sont disponibles pour la fusion
        if not all_files:
            logger.warning("Aucun fichier CSV trouvé dans le répertoire. La fusion est annulée.")
            return  # Arrête l'exécution si aucun fichier CSV n'est trouvé

        logger.info(f"{len(all_files)} fichiers CSV trouvés pour la fusion.")  # Log du nombre de fichiers trouvés

        # 🧩 Étape 2 : Lire et concaténer tous les fichiers CSV
        df_list = []  # Initialisation d'une liste pour stocker les DataFrames
        for file in all_files:
            try:
                # Lecture du fichier CSV en tant que DataFrame Pandas
                df = pd.read_csv(file)
                df_list.append(df)  # Ajoute le DataFrame à la liste
                logger.info(f"Fichier chargé avec succès : {file} (lignes : {len(df)})")  # Log du succès de la lecture
            except Exception as e:
                # Capture et log des erreurs lors de la lecture des fichiers
                logger.error(f"Erreur lors de la lecture du fichier {file} : {e}")

        # Vérifie si au moins un DataFrame valide a été collecté
        if not df_list:
            logger.warning("Aucun DataFrame valide n'a été collecté. La fusion est annulée.")
            return  # Arrête l'exécution si la liste est vide

        # Fusionne tous les DataFrames en un seul
        final_df = pd.concat(df_list, ignore_index=True)
        logger.info(f"Fusion terminée avec succès : {final_df.shape[0]} lignes et {final_df.shape[1]} colonnes.")

        # 💾 Étape 3 : Sauvegarde du DataFrame fusionné dans un fichier CSV
        final_path = os.path.join(output_dir, final_filename)  # Chemin complet du fichier final
        final_df.to_csv(final_path, index=False, encoding='utf-8')  # Sauvegarde au format CSV avec encodage UTF-8
        logger.info(f"Fichier fusionné sauvegardé avec succès : {final_path}")

    except Exception as e:
        # Capture et log des erreurs inattendues pendant le processus de fusion
        logger.error(f"Erreur lors de la fusion des fichiers CSV : {e}")
````

````python
# 📂 Importation des configurations et initialisation
import sys
from pathlib import Path

# Ajouter le dossier contenant config.py au chemin système
sys.path.append(str(Path("../src").resolve()))

# Importer toutes les configurations depuis config.py
from config import *

# Initialiser les logs pour ce notebook
notebook_name = "Analyse_Mangas_Fusion"
init_notebook_logs(notebook_name)

# 💾 Chargement du fichier fusionné
final_csv_path = CONFIG["outputs_directory"] / "mangas_all.csv"
try:
    # Charger le fichier CSV fusionné en DataFrame
    final_df = pd.read_csv(final_csv_path)
    logger.info(f"Fichier chargé avec succès : {final_csv_path}")
except FileNotFoundError:
    logger.error(f"Le fichier fusionné {final_csv_path} n'a pas été trouvé.")
    raise

# 🧐 Aperçu rapide du DataFrame fusionné
logger.info("Aperçu rapide du DataFrame fusionné :")
print(final_df.head())

# 🧹 **1. Analyse rapide avec skimpy**
logger.info("Analyse rapide avec skimpy.")
skim_result = skim(final_df)
print(skim_result)

# 🔍 **2. Identification des valeurs manquantes**
logger.info("Identification des valeurs manquantes par colonne.")
missing_values = final_df.isnull().sum()
print("\nValeurs manquantes par colonne :")
print(missing_values)

# Visualisation des valeurs manquantes
missing_percentage = (missing_values / len(final_df)) * 100
missing_percentage = missing_percentage[missing_percentage > 0].sort_values(ascending=False)

if not missing_percentage.empty:
    plt.figure(figsize=(10, 6))
    sns.barplot(x=missing_percentage.values, y=missing_percentage.index, palette="viridis")
    plt.title("Pourcentage de valeurs manquantes par colonne")
    plt.xlabel("Pourcentage de valeurs manquantes (%)")
    plt.ylabel("Colonnes")
    plt.tight_layout()
    plt.show()

# 📊 **3. Top 10 des genres**
if 'Genre' in final_df.columns:
    logger.info("Analyse des genres de mangas (Top 10).")
    genre_counts = final_df['Genre'].value_counts().head(10)
    plt.figure(figsize=(12, 6))
    sns.barplot(x=genre_counts.values, y=genre_counts.index, palette="muted")
    plt.title("Top 10 des genres de mangas")
    plt.xlabel("Nombre de mangas")
    plt.ylabel("Genres")
    plt.tight_layout()
    plt.show()

# 📚 **4. Top 10 éditeurs**
if 'Éditeur' in final_df.columns:
    logger.info("Analyse des éditeurs de mangas (Top 10).")
    editor_counts = final_df['Éditeur'].value_counts().head(10)
    plt.figure(figsize=(12, 6))
    sns.barplot(x=editor_counts.values, y=editor_counts.index, palette="coolwarm")
    plt.title("Top 10 des éditeurs")
    plt.xlabel("Nombre de mangas")
    plt.ylabel("Éditeurs")
    plt.tight_layout()
    plt.show()

# 📈 **5. Top 10 des mangas avec le plus de volumes**
if 'Titre' in final_df.columns and 'Nombre de volumes' in final_df.columns:
    logger.info("Analyse des mangas avec le plus de volumes (Top 10).")
    top_mangas = final_df[['Titre', 'Nombre de volumes']].sort_values(by='Nombre de volumes', ascending=False).head(10)
    plt.figure(figsize=(12, 6))
    sns.barplot(x=top_mangas['Nombre de volumes'], y=top_mangas['Titre'], palette="viridis")
    plt.title("Top 10 des mangas avec le plus de volumes")
    plt.xlabel("Nombre de volumes")
    plt.ylabel("Titre du manga")
    plt.tight_layout()
    plt.show()

# 🖋️ **6. Manga avec les résumés les plus longs (Top 10)**
if 'Titre' in final_df.columns and 'Résumé' in final_df.columns:
    logger.info("Analyse des mangas avec les résumés les plus longs (Top 10).")
    final_df['Longueur Résumé'] = final_df['Résumé'].fillna("").apply(len)
    top_resumes = final_df[['Titre', 'Longueur Résumé']].sort_values(by='Longueur Résumé', ascending=False).head(10)
    plt.figure(figsize=(12, 6))
    sns.barplot(x=top_resumes['Longueur Résumé'], y=top_resumes['Titre'], palette="cividis")
    plt.title("Top 10 des mangas avec les résumés les plus longs")
    plt.xlabel("Longueur du résumé (caractères)")
    plt.ylabel("Titre du manga")
    plt.tight_layout()
    plt.show()

# 📅 **7. Analyse des mangas par année**
if 'Année' in final_df.columns:
    logger.info("Analyse des mangas publiés par année.")
    year_counts = final_df['Année'].value_counts().sort_index()
    plt.figure(figsize=(12, 6))
    sns.lineplot(x=year_counts.index, y=year_counts.values, marker="o")
    plt.title("Nombre de mangas publiés par année")
    plt.xlabel("Année")
    plt.ylabel("Nombre de mangas")
    plt.tight_layout()
    plt.show()

logger.info("Analyse complète du DataFrame fusionné terminée.")
````
